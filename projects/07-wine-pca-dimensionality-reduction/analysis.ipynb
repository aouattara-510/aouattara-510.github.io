# Wine PCA & Dimensionality Reduction


```python
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt

w=load_wine(as_frame=True)
X=StandardScaler().fit_transform(w.data)
pca=PCA(n_components=2)
Z=pca.fit_transform(X)
print("Explained variance:", pca.explained_variance_ratio_, "Total:", pca.explained_variance_ratio_.sum())
plot=pd.DataFrame(Z,columns=["PC1","PC2"]); plot["class"]=w.target
plot.head()
```

```python
plt.figure(figsize=(7,5))
for c,g in plot.groupby("class"):
    plt.scatter(g.PC1,g.PC2,label=f"class {c}",alpha=.7)
plt.xlabel("PC1"); plt.ylabel("PC2"); plt.title("Wine Dataset in Two Principal Components"); plt.legend(); plt.show()
```

## Takeaway
PCA reduces dimensionality while preserving as much variance as possible; components are combinations of the original standardized features.
