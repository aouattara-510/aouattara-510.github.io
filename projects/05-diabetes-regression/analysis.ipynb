# Diabetes Progression Regression
A reproducible regression comparison using a built-in dataset.


```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pandas as pd
import numpy as np

d=load_diabetes(as_frame=True)
X=d.data; y=d.target
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42)
models={"Linear Regression":LinearRegression(),"Random Forest":RandomForestRegressor(n_estimators=250,random_state=42)}
rows=[]
for name,m in models.items():
    m.fit(X_train,y_train)
    p=m.predict(X_test)
    rows.append([name,mean_absolute_error(y_test,p),np.sqrt(mean_squared_error(y_test,p)),r2_score(y_test,p)])
pd.DataFrame(rows,columns=["Model","MAE","RMSE","R2"])
```

## Takeaway
The project compares error metrics and demonstrates why regression model selection should consider both predictive performance and interpretability.
