# Time-Series Demand Forecasting
Synthetic weekly data make the project completely self-contained.


```python
import numpy as np, pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import matplotlib.pyplot as plt

rng=np.random.default_rng(42)
n=160
t=np.arange(n)
demand=100 + .25*t + 12*np.sin(2*np.pi*t/12) + rng.normal(0,4,n)
df=pd.DataFrame({"week":t,"demand":demand})
for lag in [1,2,3,12]:
    df[f"lag_{lag}"]=df["demand"].shift(lag)
df=df.dropna().reset_index(drop=True)
train=df.iloc[:-24]; test=df.iloc[-24:]
features=[c for c in df.columns if c.startswith("lag_")]
m=RandomForestRegressor(n_estimators=300,random_state=42)
m.fit(train[features],train["demand"])
pred=m.predict(test[features])
print("MAE:", round(mean_absolute_error(test["demand"],pred),2))
```

```python
plt.figure(figsize=(8,4))
plt.plot(test["week"],test["demand"],label="Actual")
plt.plot(test["week"],pred,label="Forecast")
plt.title("Synthetic Demand Forecast"); plt.xlabel("Week"); plt.ylabel("Demand"); plt.legend(); plt.show()
```

## Takeaway
Forecast validation should respect time order. A production project would use rolling validation and compare against simple seasonal baselines.
