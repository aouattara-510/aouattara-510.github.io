# A/B Testing Simulation
Synthetic conversion counts demonstrate hypothesis testing without external data.


```python
import numpy as np
from math import sqrt
from statistics import NormalDist

n_a,n_b=2000,2000
conv_a,conv_b=210,250
p_a,p_b=conv_a/n_a,conv_b/n_b
p_pool=(conv_a+conv_b)/(n_a+n_b)
se=sqrt(p_pool*(1-p_pool)*(1/n_a+1/n_b))
z=(p_b-p_a)/se
p_value=2*(1-NormalDist().cdf(abs(z)))
print({"rate_A":p_a,"rate_B":p_b,"lift":p_b-p_a,"z":z,"p_value":p_value})
```

## Takeaway
Statistical significance is only one consideration; experiment design, sample size, practical effect size, and business costs also matter.
