# SQL Sales Analytics
This notebook creates an in-memory SQLite database so it runs anywhere Python is available.


```python
import sqlite3, pandas as pd
con=sqlite3.connect(":memory:")
customers=pd.DataFrame({"customer_id":[1,2,3,4],"region":["East","West","East","South"]})
orders=pd.DataFrame({"order_id":[101,102,103,104,105,106],"customer_id":[1,1,2,3,4,2],"revenue":[120,250,90,310,175,220]})
customers.to_sql("customers",con,index=False)
orders.to_sql("orders",con,index=False)
pd.read_sql_query('''
SELECT c.region,
       COUNT(o.order_id) AS orders,
       ROUND(SUM(o.revenue),2) AS revenue,
       ROUND(AVG(o.revenue),2) AS avg_order
FROM orders o
JOIN customers c USING(customer_id)
GROUP BY c.region
ORDER BY revenue DESC
''',con)
```

## Takeaway
SQL is often the simplest and most efficient tool when the business question can be answered directly from structured relational data.
