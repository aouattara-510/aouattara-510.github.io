# Customer Segmentation with K-Means
Synthetic data keep this notebook portable and reproducible.


```python
import numpy as np, pandas as pd
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X,_=make_blobs(n_samples=500, centers=4, n_features=3, cluster_std=1.2, random_state=42)
df=pd.DataFrame(X,columns=["annual_spend","visits","avg_order_value"])
scaled=StandardScaler().fit_transform(df)
scores={}
for k in range(2,7):
    labels=KMeans(n_clusters=k,n_init=20,random_state=42).fit_predict(scaled)
    scores[k]=silhouette_score(scaled,labels)
scores
```

```python
best_k=max(scores,key=scores.get)
model=KMeans(n_clusters=best_k,n_init=20,random_state=42)
df["segment"]=model.fit_predict(scaled)
df.groupby("segment").mean().round(2)
```

## Takeaway
Cluster labels are descriptive segments, not inherent customer types. Business interpretation should be validated before action.
