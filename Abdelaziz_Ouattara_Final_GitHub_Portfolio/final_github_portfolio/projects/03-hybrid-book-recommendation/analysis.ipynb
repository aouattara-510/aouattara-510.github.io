# Hybrid Book Recommendation Demo
This notebook demonstrates a small hybrid recommender using content similarity plus reader-rating similarity. The sample data are synthetic so the notebook runs without external downloads.


```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

books = pd.DataFrame({
    "book_id":[1,2,3,4,5,6],
    "title":["Data Stories","Machine Learning Basics","Python for Analysis","Modern Databases","AI Ethics","Applied Statistics"],
    "author":["A","B","C","D","E","F"],
    "genre":["data storytelling","machine learning","python data","sql databases","ai ethics","statistics analysis"],
    "summary":[
        "visual communication and storytelling with data",
        "supervised learning classification and regression",
        "python pandas analysis visualization",
        "relational databases sql analytics",
        "responsible artificial intelligence fairness transparency",
        "probability inference regression experiments"
    ]
})
books
```

```python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform((books["genre"] + " " + books["summary"]))
content_sim = cosine_similarity(X)
content_sim.round(2)
```

```python
ratings = pd.DataFrame({
    1:[5,4,0,0],
    2:[5,0,4,0],
    3:[4,5,0,0],
    4:[0,0,4,5],
    5:[0,3,5,4],
    6:[3,4,0,5],
}, index=["reader_1","reader_2","reader_3","reader_4"]).T
ratings.index.name="book_id"
ratings
```

```python
def recommend_similar(title, n=3):
    idx = books.index[books["title"].eq(title)][0]
    scores = list(enumerate(content_sim[idx]))
    ranked = sorted(scores, key=lambda x:x[1], reverse=True)[1:n+1]
    return books.iloc[[i for i,_ in ranked]][["title","genre"]]

recommend_similar("Machine Learning Basics")
```

## Interpretation
A production hybrid recommender would combine content similarity with user-item interaction signals, tune weights, and evaluate relevance, diversity, novelty, cold-start behavior, and popularity bias.
