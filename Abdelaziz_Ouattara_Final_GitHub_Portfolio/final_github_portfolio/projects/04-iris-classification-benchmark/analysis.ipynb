# Iris Classification Benchmark
This project compares common classifiers on a clean multiclass dataset.


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import pandas as pd

data=load_iris()
X=pd.DataFrame(data.data, columns=data.feature_names)
y=data.target
X.head()
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.25,random_state=42,stratify=y)
models={
 "Logistic Regression":LogisticRegression(max_iter=1000),
 "Decision Tree":DecisionTreeClassifier(random_state=42),
 "Random Forest":RandomForestClassifier(n_estimators=200,random_state=42)
}
rows=[]
for name,m in models.items():
    cv=cross_val_score(m,X_train,y_train,cv=5).mean()
    m.fit(X_train,y_train)
    pred=m.predict(X_test)
    rows.append([name,cv,accuracy_score(y_test,pred)])
pd.DataFrame(rows,columns=["Model","CV Accuracy","Test Accuracy"])
```

## Takeaway
The notebook demonstrates consistent train/test splitting, cross-validation, model comparison, and concise interpretation.
